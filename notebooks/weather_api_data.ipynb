The subtask requires loading two CSV files, parsing the 'datetime' column as datetime objects, creating a complete date range for 2025, reindexing `df_2025` to include all days of the year, and then concatenating `df_2024` and the reindexed `df_2025_full` into a single DataFrame.
````python
import pandas as pd

# 1. Load 'dindigul_weather_2024.csv'
df_2024 = pd.read_csv('dindigul_weather_2024.csv', parse_dates=['datetime'])
print(f"Loaded {len(df_2024)} records for 2024.")

# 2. Load 'dindigul_weather_2025.csv'
df_2025 = pd.read_csv('dindigul_weather_2025.csv', parse_dates=['datetime'])
print(f"Loaded {len(df_2025)} records for 2025 (may be partial).")

# 3. Create a complete date range for the year 2025
full_date_range_2025 = pd.date_range(start='2025-01-01', end='2025-12-31', freq='D')

# 4. Reindex df_2025 using this complete date range
df_2025_full = df_2025.set_index('datetime').reindex(full_date_range_2025).reset_index()
df_2025_full = df_2025_full.rename(columns={'index': 'datetime'})
print(f"Full 2025 DataFrame has {len(df_2025_full)} entries after reindexing.")

# 5. Combine df_2024 and df_2025_full
df_combined = pd.concat([df_2024, df_2025_full], ignore_index=True)

print("\nCombined DataFrame head:")
print(df_combined.head())
print("\nCombined DataFrame info:")
df_combined.info()
````

## Feature Engineering for ML Imputation

### Subtask:
Extract relevant time-based features such as 'month', 'day_of_year', and 'day_of_week' from the 'datetime' column. Convert 'sunrise' and 'sunset' time strings into a numerical format (e.g., minutes/seconds from midnight) to allow for regression modeling.


**Reasoning**:
The subtask requires creating new time-based features and converting time strings to numerical format. This code block will extract month, day of year, and day of week from the 'datetime' column and convert 'sunrise' and 'sunset' times to seconds from midnight, handling potential missing values.



````python
import numpy as np

# 1. Extract time-based features from 'datetime'
df_combined['month'] = df_combined['datetime'].dt.month
df_combined['day_of_year'] = df_combined['datetime'].dt.dayofyear
df_combined['day_of_week'] = df_combined['datetime'].dt.dayofweek # Monday=0, Sunday=6

# Function to convert 'HH:MM:SS' string to total seconds from midnight
def time_to_seconds(time_str):
    if pd.isna(time_str):
        return np.nan
    try:
        hours, minutes, seconds = map(int, time_str.split(':'))
        return hours * 3600 + minutes * 60 + seconds
    except ValueError:
        return np.nan # Handle cases where time_str is not in expected format

# 2. Convert 'sunrise' and 'sunset' to numerical format (seconds from midnight)
df_combined['sunrise_seconds'] = df_combined['sunrise'].apply(time_to_seconds)
df_combined['sunset_seconds'] = df_combined['sunset'].apply(time_to_seconds)

print("New features added to df_combined:")
print(df_combined[['datetime', 'month', 'day_of_year', 'day_of_week', 'sunrise', 'sunrise_seconds', 'sunset', 'sunset_seconds']].head())
print(df_combined[['month', 'day_of_year', 'day_of_week', 'sunrise_seconds', 'sunset_seconds']].info())
````

## Impute Numerical Missing Values using Machine Learning

### Subtask:
Use a regression model (e.g., RandomForestRegressor) to impute the missing numerical values for 2025. Iterate through each numerical target column, train the model on the 2024 data (and complete 2025 data), and then predict the missing values for the days in 2025 that have `NaN`.

#### Instructions
1.  Filter `df_combined` to create a training set (`df_train`) consisting of all rows where the numerical target columns are *not* null. This will primarily be data from 2024 and the complete days in 2025.
2.  Filter `df_combined` to create a prediction set (`df_predict`) consisting of rows where the numerical target columns *are* null (i.e., the missing days in 2025).
3.  For each numerical column in `target_columns` (excluding 'conditions' and 'description'):
    a.  Define the target variable (`y`) and features (`X`) from `df_train`.

**Reasoning**:
The subtask requires using a regression model to impute missing numerical values. This code block will implement a RandomForestRegressor to predict and fill missing numerical data in the identified target columns for 2025, using the available 2024 data and complete 2025 data for training.



````python
from sklearn.ensemble import RandomForestRegressor

# Filter out non-numerical target columns for ML imputation
numerical_target_columns = [col for col in target_columns if col not in ['sunrise', 'sunset', 'conditions', 'description']]

# Prepare training and prediction sets
df_train = df_combined[df_combined[numerical_target_columns].notna().all(axis=1)].copy()
df_predict = df_combined[df_combined[numerical_target_columns].isna().any(axis=1)].copy()

print(f"Training data size: {len(df_train)}")
print(f"Prediction data size (missing values): {len(df_predict)}")

# Impute each numerical target column
for col in numerical_target_columns:
    # Define features (X) and target (y) for the current column
    X_train = df_train[feature_columns]
    y_train = df_train[col]

    X_predict = df_predict[feature_columns]

    # Initialize and train RandomForestRegressor
    model = RandomForestRegressor(n_estimators=100, random_state=42)
    model.fit(X_train, y_train)

    # Predict missing values
    predicted_values = model.predict(X_predict)

    # Fill NaN values in df_combined for the current column
    # We need to ensure we fill only the NaNs in the prediction set's rows
    df_combined.loc[df_predict.index, col] = predicted_values
    print(f"Imputed missing values for column: {col}")

# 4. Update 'datetimeEpoch' based on 'datetime' for consistency if it was imputed numerically
# Assuming datetimeEpoch is seconds since epoch for the start of the day (midnight UTC)
# Convert datetime to UTC timestamp and then to integer seconds
# Only update if datetimeEpoch was one of the imputed columns
if 'datetimeEpoch' in numerical_target_columns:
    df_combined['datetimeEpoch'] = df_combined['datetime'].apply(lambda x: int(x.timestamp()))
    print("Updated 'datetimeEpoch' based on 'datetime' column.")

print("\nCombined DataFrame after numerical imputation:")
print(df_combined.info())
print(df_combined.head())
````

**Reasoning**:
After imputing both numerical and categorical values, the final step for this subtask is to convert the imputed numerical `sunrise_seconds` and `sunset_seconds` back into 'HH:MM:SS' string format for the 'sunrise' and 'sunset' columns, extract the 2025 data, and then save the complete `dindigul_weather_2025.csv` file with the specified columns.



````python
from datetime import time

# Function to convert total seconds from midnight back to 'HH:MM:SS' string
def seconds_to_time_str(total_seconds):
    if pd.isna(total_seconds):
        return np.nan
    total_seconds = int(total_seconds)
    hours = total_seconds // 3600
    minutes = (total_seconds % 3600) // 60
    seconds = total_seconds % 60
    return f'{hours:02d}:{minutes:02d}:{seconds:02d}'

# Convert imputed 'sunrise_seconds' and 'sunset_seconds' back to 'HH:MM:SS' strings
df_combined['sunrise'] = df_combined['sunrise_seconds'].apply(seconds_to_time_str)
df_combined['sunset'] = df_combined['sunset_seconds'].apply(seconds_to_time_str)

# Define the list of desired columns for the output CSV
desired_columns_output = [
    'datetime', 'datetimeEpoch', 'temp', 'humidity', 'windspeed',
    'solarradiation', 'sunrise', 'sunriseEpoch', 'sunset', 'sunsetEpoch',
    'conditions', 'description'
]

# Filter for year 2025 from the fully imputed combined DataFrame
df_2025_imputed = df_combined[df_combined['datetime'].dt.year == 2025].copy()

# Select only the desired columns
df_2025_imputed = df_2025_imputed[desired_columns_output]

# Convert 'datetime' column back to 'YYYY-MM-DD' string format for CSV saving
df_2025_imputed['datetime'] = df_2025_imputed['datetime'].dt.strftime('%Y-%m-%d')

# Save the completed df_2025 to a CSV file
df_2025_imputed.to_csv('dindigul_weather_2025.csv', index=False)

print(f"Saved {len(df_2025_imputed)} records for 2025 to 'dindigul_weather_2025.csv' after imputation.")
print("First 5 rows of the imputed 2025 data:")
print(df_2025_imputed.head())
print("Last 5 rows of the imputed 2025 data (including imputed days):")
print(df_2025_imputed.tail())
````
